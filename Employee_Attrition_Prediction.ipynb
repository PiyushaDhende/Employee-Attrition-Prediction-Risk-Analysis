Step 10 — Train Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression(class_weight='balanced',random_state=42,max_iter=1000)
log_model.fit(X_train_scaled,y_train)
```

Step 11 — Make predictions

```python
y_pred = log_model.predict(X_test_scaled)
y_prob = log_model.predict_proba(X_test_scaled)[:,1]
```

Step 12 — Evaluate the model

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    class_weight='balanced',
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1 Score:", f1_score(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```

Logistic Regression → 65.96%
Random Forest → 10.64%
That means Random Forest is missing almost 90% of the employees who actually leave.


Logistic Regression is more useful for the actual attrition-detection objective, despite its lower accuracy.
Random Forest has a much better ROC-AUC (80.75%), though, which tells us it has useful ranking/discrimination ability. The problem is its default 0.5 classification threshold isn't giving us enough positive predictions.

Step 17 — Improve Random Forest

```python
rf_model_tuned = RandomForestClassifier(
    n_estimators=300,
    class_weight={0: 1, 1: 3},
    random_state=42
)

rf_model_tuned.fit(X_train, y_train)
```

```python
y_pred_rf_tuned = rf_model_tuned.predict(X_test)
y_prob_rf_tuned = rf_model_tuned.predict_proba(X_test)[:, 1]
```

Step 19 — Evaluate the tuned Random Forest

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf_tuned))
print("Precision:", precision_score(y_test, y_pred_rf_tuned))
print("Recall:", recall_score(y_test, y_pred_rf_tuned))
print("F1 Score:", f1_score(y_test, y_pred_rf_tuned))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf_tuned))
```
